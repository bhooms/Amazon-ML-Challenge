## Normalization

```python
import re

def normalize_name(name):
    name = str(name).lower()
    
    # Remove punctuation
    name = re.sub(r"[^a-z0-9\s]", " ", name)
    
    # Remove extra spaces
    name = re.sub(r"\s+", " ", name).strip()
    
    return name
```

## Cleaning buisness adresses

```python
def normalize_address(address):
    address = str(address).lower()

    # Remove punctuation
    address = re.sub(r"[^a-z0-9\s]", " ", address)

    # Common address abbreviations
    address = re.sub(r"\bstreet\b", "st", address)
    address = re.sub(r"\broad\b", "rd", address)
    address = re.sub(r"\bavenue\b", "ave", address)
    address = re.sub(r"\broadway\b", "rd", address)
    address = re.sub(r"\bbuilding\b", "bldg", address)
    address = re.sub(r"\bapartment\b", "apt", address)
    address = re.sub(r"\bsuite\b", "ste", address)

    # Remove extra spaces
    address = re.sub(r"\s+", " ", address).strip()

    return address
```

```python
train_source1["address_clean"] = train_source1["business_address"].apply(normalize_address)
train_source2["address_clean"] = train_source2["business_address"].apply(normalize_address)
train_source3["address_clean"] = train_source3["business_address"].apply(normalize_address)

print(
    train_source1[
        ["business_address", "address_clean"]
    ].head(10).to_string(index=False)
)
```

## Feature Table

```python
def create_features(s1_row, candidate_row):
    name_sim = text_similarity(
        s1_row["name_clean"],
        candidate_row["name_clean"]
    )

    address_sim = text_similarity(
        s1_row["address_clean"],
        candidate_row["address_clean"]
    )

    country_match = int(
        s1_row["country"] == candidate_row["country"]
    )

    return {
        "name_similarity": name_sim,
        "address_similarity": address_sim,
        "country_match": country_match
    }
```

## Adding more better features

```python
def length_similarity(a, b):
    a = str(a)
    b = str(b)

    max_len = max(len(a), len(b))

    if max_len == 0:
        return 1.0

    return 1 - abs(len(a) - len(b)) / max_len
```

```python
training_pairs["name_length_similarity"] = [
    length_similarity(a, b)
    for a, b in zip(
        training_pairs["s1_name"],
        training_pairs["candidate_name"]
    )
]
```

```python
training_pairs["address_length_similarity"] = [
    length_similarity(a, b)
    for a, b in zip(
        training_pairs["s1_address"],
        training_pairs["candidate_address"]
    )
]
```

```python
training_pairs["exact_name_match"] = (
    training_pairs["s1_name"] ==
    training_pairs["candidate_name"]
).astype(int)
```

## Adding token overlap

```python
def token_overlap(a, b):
    tokens_a = set(str(a).split())
    tokens_b = set(str(b).split())

    if not tokens_a or not tokens_b:
        return 0.0

    return len(tokens_a & tokens_b) / len(tokens_a | tokens_b)
```

```python
training_pairs["name_token_overlap"] = [
    token_overlap(a, b)
    for a, b in zip(
        training_pairs["s1_name"],
        training_pairs["candidate_name"]
    )
]
```

## Address blocking rule

```python
def address_block_key(address):
    tokens = str(address).split()

    if not tokens:
        return ""

    return tokens[0]
```

```python
test_candidates["address_block"] = (
    test_candidates["candidate_address"]
    .apply(address_block_key)
)
```

```python
address_index = (
    test_candidates
    .groupby("address_block")["candidate_id"]
    .apply(list)
    .to_dict()
)
```

```python
def get_candidates_full(
    s1_name,
    s1_country,
    s1_address
):
    candidates = set()

    # Block 1: exact name
    candidates.update(
        name_index.get(s1_name, [])
    )

    # Block 2: country + name prefix
    candidates.update(
        prefix_index.get(
            name_prefix_key(
                s1_name,
                s1_country
            ),
            []
        )
    )

    # Block 3: address token
    address_key = address_block_key(
        s1_address
    )

    if address_key:
        candidates.update(
            address_index.get(
                address_key,
                []
            )
        )

    return list(candidates)
```

## Generating candidate_pairs.tsv

```python
def compact_name(name):
    name = str(name).lower()
    return re.sub(r"[^a-z0-9]", "", name)


def address_number_key(address):
    address = str(address).lower()

    numbers = re.findall(r"\d+", address)

    if numbers:
        return numbers[0]

    return ""


def name_prefix_key_7(name, country):
    compact = compact_name(name)

    return (
        str(country).lower(),
        compact[:7]
    )


def name_prefix_length_key(name, country):
    compact = compact_name(name)

    length_bucket = len(compact) // 4

    return (
        str(country).lower(),
        compact[:4],
        length_bucket
    )


def address_country_key(address, country):
    number = address_number_key(address)

    if number:
        return (
            str(country).lower(),
            number
        )

    return None
```

```python
test_candidates["compact_name"] = (
    test_candidates["candidate_name"]
    .apply(compact_name)
)

test_candidates["block_name_7"] = [
    name_prefix_key_7(name, country)
    for name, country in zip(
        test_candidates["candidate_name"],
        test_candidates["candidate_country"]
    )
]

test_candidates["block_name_length"] = [
    name_prefix_length_key(name, country)
    for name, country in zip(
        test_candidates["candidate_name"],
        test_candidates["candidate_country"]
    )
]

test_candidates["block_address_country"] = [
    address_country_key(address, country)
    for address, country in zip(
        test_candidates["candidate_address"],
        test_candidates["candidate_country"]
    )
]

print("Blocking columns created.")
```

```python
exact_name_index = (
    test_candidates
    .groupby("candidate_name")["candidate_id"]
    .apply(list)
    .to_dict()
)

name7_index = (
    test_candidates
    .groupby("block_name_7")["candidate_id"]
    .apply(list)
    .to_dict()
)

name_length_index = (
    test_candidates
    .groupby("block_name_length")["candidate_id"]
    .apply(list)
    .to_dict()
)

address_country_index = (
    test_candidates
    .dropna(subset=["block_address_country"])
    .groupby("block_address_country")["candidate_id"]
    .apply(list)
    .to_dict()
)

print("Indexes created.")
```

```python
# ============================================================
# Final candidate blocking function
# ============================================================

MAX_NAME7_BLOCK = 1000
MAX_NAME_LENGTH_BLOCK = 1000
MAX_ADDRESS_COUNTRY_BLOCK = 300


def get_tight_candidates(name, country, address):

    candidates = set()

    # --------------------------------------------------------
    # Block 1: exact normalized name
    # --------------------------------------------------------

    exact = exact_name_index.get(name, [])

    candidates.update(exact)

    # --------------------------------------------------------
    # Block 2: country + first 7 characters of name
    # --------------------------------------------------------

    key7 = name_prefix_key_7(
        name,
        country
    )

    block7 = name7_index.get(
        key7,
        []
    )

    if len(block7) <= MAX_NAME7_BLOCK:
        candidates.update(block7)

    # --------------------------------------------------------
    # Block 3: country + first 4 characters + length bucket
    # --------------------------------------------------------

    key_length = name_prefix_length_key(
        name,
        country
    )

    block_length = name_length_index.get(
        key_length,
        []
    )

    if len(block_length) <= MAX_NAME_LENGTH_BLOCK:
        candidates.update(block_length)

    # --------------------------------------------------------
    # Block 4: country + address number
    # --------------------------------------------------------

    address_key = address_country_key(
        address,
        country
    )

    if address_key is not None:

        address_block = address_country_index.get(
            address_key,
            []
        )

        if len(address_block) <= MAX_ADDRESS_COUNTRY_BLOCK:
            candidates.update(address_block)

    return candidates
```

## Scoring candidate pairs

```python
import os
import gc
import pandas as pd
from difflib import SequenceMatcher

scored_file = "output/scored_candidates.tsv"

if os.path.exists(scored_file):
    os.remove(scored_file)


def similarity_series(col1, col2):

    return [
        SequenceMatcher(
            None,
            str(a),
            str(b)
        ).ratio()

        for a, b in zip(
            col1,
            col2
        )
    ]


score_chunk_size = 2000

first_score_chunk = True

total_scored = 0


for chunk in pd.read_csv(
    candidate_file,
    sep="\t",
    chunksize=score_chunk_size
):

    chunk = chunk[
        chunk["candidate_entity_ids"]
        .fillna("")
        .str.strip()
        != ""
    ].copy()

    if len(chunk) == 0:
        continue

    # Expand comma-separated candidates
    chunk["candidate_id"] = (
        chunk["candidate_entity_ids"]
        .str.split(",")
    )

    pairs = (
        chunk[
            [
                "source1_entity_id",
                "candidate_id"
            ]
        ]
        .explode("candidate_id")
    )

    pairs["candidate_id"] = (
        pairs["candidate_id"]
        .str.strip()
    )

    # Add S1 information
    pairs = pairs.merge(
        s1_test_lookup,
        left_on="source1_entity_id",
        right_on="s1_id",
        how="left"
    )

    # Add candidate information
    pairs = pairs.merge(
        candidate_lookup,
        left_on="candidate_id",
        right_index=True,
        how="left"
    )

    # --------------------------------------------------------
    # Same features used by Model 3
    # --------------------------------------------------------

    pairs["name_similarity"] = similarity_series(
        pairs["s1_name"],
        pairs["candidate_name"]
    )

    pairs["address_similarity"] = similarity_series(
        pairs["s1_address"],
        pairs["candidate_address"]
    )

    pairs["country_match"] = (
        pairs["s1_country"]
        ==
        pairs["candidate_country"]
    ).astype(int)

    pairs["name_length_similarity"] = [
        length_similarity(a, b)
        for a, b in zip(
            pairs["s1_name"],
            pairs["candidate_name"]
        )
    ]

    pairs["address_length_similarity"] = [
        length_similarity(a, b)
        for a, b in zip(
            pairs["s1_address"],
            pairs["candidate_address"]
        )
    ]

    pairs["exact_name_match"] = (
        pairs["s1_name"]
        ==
        pairs["candidate_name"]
    ).astype(int)

    pairs["name_token_overlap"] = [
        token_overlap(a, b)
        for a, b in zip(
            pairs["s1_name"],
            pairs["candidate_name"]
        )
    ]

    X_test = pairs[
        feature_columns_v2
    ]

    pairs["match_probability"] = (
        model_v3
        .predict_proba(X_test)[:, 1]
    )

    save_columns = [
        "source1_entity_id",
        "candidate_id",
        "match_probability"
    ]

    pairs[
        save_columns
    ].to_csv(
        scored_file,
        sep="\t",
        index=False,
        mode="w" if first_score_chunk else "a",
        header=first_score_chunk
    )

    first_score_chunk = False

    total_scored += len(pairs)

    print(
        f"Scored candidate pairs: "
        f"{total_scored:,}"
    )

    del pairs
    del X_test

    gc.collect()


print()
print("Scoring completed.")
print("Total scored:", total_scored)
```

## Checking if matches are actually candidates

```python
matching_dict = dict(
    zip(
        matching_results[
            "source1_entity_id"
        ],
        matching_results[
            "matched_entity_ids"
        ]
    )
)

violations = []

for chunk in pd.read_csv(
    candidate_file,
    sep="\t",
    chunksize=10000
):

    for _, row in chunk.iterrows():

        s1_id = row[
            "source1_entity_id"
        ]

        matches = str(
            matching_dict.get(
                s1_id,
                ""
            )
        ).strip()

        if not matches:
            continue

        candidate_string = str(
            row[
                "candidate_entity_ids"
            ]
        )

        candidate_set = set(
            x.strip()
            for x in candidate_string.split(",")
            if x.strip()
        )

        match_set = set(
            x.strip()
            for x in matches.split(",")
            if x.strip()
        )

        missing = (
            match_set
            -
            candidate_set
        )

        if missing:

            violations.append({
                "source1_entity_id":
                    s1_id,
                "missing_matches":
                    ",".join(
                        sorted(missing)
                    )
            })

            if len(violations) >= 10:
                break

    if len(violations) >= 10:
        break


print(
    "Violations found:",
    len(violations)
)

if violations:
    print(
        pd.DataFrame(violations)
    )
else:
    print(
        "Every predicted match belongs "
        "to its candidate set."
    )
```
