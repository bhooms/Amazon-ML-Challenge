## Inspecting one actual buisness

```python
s1_id = "S1-965667"

s1_business = train_source1[
    train_source1["entity_id"] == s1_id
]

print(s1_business.to_string(index=False))
```

```python
matches = ground_truth[
    ground_truth["source1_entity_id"] == s1_id
]["matched_entity_ids"].iloc[0]

print("Correct matches:")
print(matches)
```

## Finding s2/s3 matches

```python
# Get all matching IDs as a list
match_ids = matches.split(",")

# Find those records in Source 2 and Source 3
matched_s2 = train_source2[
    train_source2["entity_id"].isin(match_ids)
]

matched_s3 = train_source3[
    train_source3["entity_id"].isin(match_ids)
]

print("SOURCE 2 MATCHES:")
print(matched_s2.to_string(index=False))

print("\nSOURCE 3 MATCHES:")
print(matched_s3.to_string(index=False))
```
